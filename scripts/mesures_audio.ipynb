### Graphique 1 - Réponse en fréquence

```python
# Création du graphique de réponse en fréquence
fig1, ax1 = plt.subplots(figsize=(10, 6))

ax1.plot(df1['Fréquence [Hz]'], df1['dBr B (réf 1 kHz)'],
         marker='s', linestyle='-', linewidth=1.5, markersize=6,
         label='Voie B', color='#d62728')
ax1.plot(df1['Fréquence [Hz]'], df1['dBr A (réf 1 kHz)'],
         marker='o', linestyle='-', linewidth=1.5, markersize=6,
         label='Voie A', color='#1f77b4')

ax1.set_xlabel('Fréquence [Hz]', fontsize=11, fontweight='bold')
ax1.set_ylabel('Niveau relatif [dB]', fontsize=11, fontweight='bold')
ax1.set_title('Réponse en fréquence, 44,1 kHz', fontsize=13, fontweight='bold', pad=15)
ax1.legend(loc='lower left', fontsize=10, framealpha=0.9)
ax1.grid(True, alpha=0.3, linestyle='--')

# Échelle logarithmique pour l'axe X
ax1.set_xscale('log')
ax1.set_xticks([10, 100, 1000, 10000])
ax1.set_xticklabels(['10', '100', '1000', '10000'])

plt.tight_layout()
plt.savefig(output_dir / "reponse_frequence.png", dpi=300, bbox_inches='tight')
plt.show()

print(f"Graphique 1 enregistré : {output_dir / 'reponse_frequence.png'}")
```

### Graphique 3 - THD+N en fonction du niveau

```python
# Création du graphique THD+N en fonction du niveau
fig3, ax3 = plt.subplots(figsize=(10, 6))

ax3.plot(df2['Niveau [dBFS]'], df2['THD+N A [dB]'],
         marker='o', linestyle='-', linewidth=1.5, markersize=6,
         label='Voie A', color='#1f77b4')
ax3.plot(df2['Niveau [dBFS]'], df2['THD+N B [dB]'],
         marker='s', linestyle='-', linewidth=1.5, markersize=6,
         label='Voie B', color='#d62728')

ax3.set_xlabel('Niveau [dBFS]', fontsize=11, fontweight='bold')
ax3.set_ylabel('THD+N [dB]', fontsize=11, fontweight='bold')
ax3.set_title('THD+N en fonction du niveau, 1 kHz', fontsize=13, fontweight='bold', pad=15)
ax3.legend(loc='lower left', fontsize=10, framealpha=0.9)
ax3.grid(True, alpha=0.3, linestyle='--')

# Inverser l'axe X pour correspondre au graphique original (de -65 à 0)
ax3.invert_xaxis()
ax3.set_xlim(-65, 0)

# Ajouter un deuxième axe Y à droite
ax3_right = ax3.twinx()
ax3_right.set_ylim(-100, -30)
ax3_right.set_yticks([-100, -90, -80, -70, -60, -50, -40, -30])

plt.tight_layout()
plt.savefig(output_dir / "thdn_niveau.png", dpi=300, bbox_inches='tight')
plt.show()

print(f"Graphique 3 enregistré : {output_dir / 'thdn_niveau.png'}")
```
